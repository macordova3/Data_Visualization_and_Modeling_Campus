# Step 5: Evidence



## 13. Create the first static Plotly figure

Build the first figure before placing it in Dash. Give it a descriptive title and human-readable axis and legend labels.



```python
# Define orders for the charts
location_order = ['Onsite', 'Hybrid', 'Remote']
stress_order = ['Low', 'Medium', 'High']

stress_counts = analysis.groupby(['Work_Location', 'Stress_Level'], as_index=False).size().rename(columns={'size': 'Cases'})

stress_fig = px.bar(
    stress_counts,
    x="Work_Location",
    y="Cases",
    color="Stress_Level",
    barmode="group",
    category_orders={"Work_Location": location_order, "Stress_Level": stress_order},
    title="Stress Levels by Work Location",
    labels={
        "Work_Location": "Work Location", 
        "Cases": "Number of Employees", 
        "Stress_Level": "Stress Level"
    },
    color_discrete_map={"Low": "blue", "Medium": "gold", "High": "red"}
)
stress_fig.show()
```

**What evidence should a viewer obtain from the first figure?**

A viewer should clearly see how the total number of employees reporting Low, Medium, or High stress is distributed across the three work locations. Using a grouped bar chart makes it easy to compare the volume of high stress cases in a remote setting versus an onsite setting.


## 14. Create the second static Plotly figure

The second figure should add different but related evidence rather than repeat the first figure in another style.



```python
# Calculate the average social isolation rating for each location
isolation_avg = analysis.groupby('Work_Location', as_index=False)['Social_Isolation_Rating'].mean()

isolation_fig = px.bar(
    isolation_avg,
    x="Work_Location",
    y="Social_Isolation_Rating",
    color="Work_Location",
    category_orders={"Work_Location": location_order},
    title="Average Social Isolation Rating by Work Location",
    labels={
        "Work_Location": "Work Location", 
        "Social_Isolation_Rating": "Avg Isolation Rating (1-5)"
    },
    color_discrete_map={"Onsite": "steelblue", "Hybrid": "darkorange", "Remote": "mediumpurple"}
)
# Lock the y axis to the max rating of 5
isolation_fig.update_layout(yaxis=dict(range=[0, 5]), showlegend=False)
isolation_fig.show()
```

**What does the second figure add to the first figure's evidence?**

The second figure adds a quantitative metric (average isolation on a 1-5 scale) to complement the categorical stress data. It reveals whether the lack of a physical office drives up feelings of isolation, providing a counter-balance to the stress metrics.

## 16. Create reusable figure code

Write a function that accepts the selected control value, filters or summarizes `analysis`, creates both Plotly figures, and returns the figures in a consistent order. Fix axis ranges or category orders when they need to remain stable across selections.



```python
# Calculate y-axis maximums so the charts don't bounce around when filtering
max_all = analysis.groupby(['Work_Location', 'Stress_Level']).size().max() * 1.1
max_industry = analysis.groupby(['Industry', 'Work_Location', 'Stress_Level']).size().max() * 1.1

def make_dashboard_figures(selected_industry):
    # Filter the dataset based on the dropdown selection
    if selected_industry == 'All Industries':
        filtered_df = analysis
        y_max_stress = max_all
    else:
        filtered_df = analysis[analysis['Industry'] == selected_industry]
        y_max_stress = max_industry

    # Chart 1: Stress Levels
    stress_counts = filtered_df.groupby(['Work_Location', 'Stress_Level'], as_index=False).size().rename(columns={'size': 'Cases'})
    fig1 = px.bar(
        stress_counts,
        x="Work_Location",
        y="Cases",
        color="Stress_Level",
        barmode="group",
        category_orders={"Work_Location": location_order, "Stress_Level": stress_order},
        title=f"Stress Levels in {selected_industry}",
        labels={"Work_Location": "Work Location", "Cases": "Employees", "Stress_Level": "Stress"},
        color_discrete_map={"Low": "blue", "Medium": "gold", "High": "red"},
        range_y=[0, y_max_stress]
    )

    # Chart 2: Social Isolation
    isolation_avg = filtered_df.groupby('Work_Location', as_index=False)['Social_Isolation_Rating'].mean()
    fig2 = px.bar(
        isolation_avg,
        x="Work_Location",
        y="Social_Isolation_Rating",
        color="Work_Location",
        category_orders={"Work_Location": location_order},
        title=f"Avg Isolation in {selected_industry}",
        labels={"Work_Location": "Work Location", "Social_Isolation_Rating": "Avg Isolation (1-5)"},
        color_discrete_map={"Onsite": "steelblue", "Hybrid": "darkorange", "Remote": "mediumpurple"},
        range_y=[0, 5]
    )
    fig2.update_layout(showlegend=False)
    
    return fig1, fig2
```

## 17. Build the layout

Create a Dash app and its layout. Include the title, purpose statement, interactive control, two empty `dcc.Graph` components with unique IDs, and short conclusion and limitation statements.



```python
app = Dash(__name__)

app.layout = html.Div([
    html.H1("Impact of Remote Work on Employee Well-being"),
    html.P("This dashboard explores how work location affects employee stress and social isolation across different industries."),
    
    html.Label("Select an Industry:", htmlFor="industry-dropdown"),
    dcc.Dropdown(
        id="industry-dropdown",
        options=industry_options,
        value="All Industries",
        clearable=False
    ),
    
    html.Div([
        dcc.Graph(id="stress-graph", style={'flex': '1', 'minWidth': '400px'}),
        dcc.Graph(id="isolation-graph", style={'flex': '1', 'minWidth': '400px'})
    ], style={'display': 'flex', 'flexWrap': 'wrap', 'gap': '20px', 'marginTop': '20px'}),
    
    html.H3("Conclusion"),
    html.P("The data reveals that average social isolation remains strictly moderate (around 3 out of 5) regardless of work location. Furthermore, the distribution of high stress remains persistent across Remote, Hybrid, and Onsite environments. Eliminating the office commute does not automatically eliminate high workplace stress."),
    
    html.H3("Limitation"),
    html.P("This dataset relies on self-reported survey responses collected at a single point in time. It reveals correlations between work location and stress, but cannot prove direct causation, and self-reported metrics are inherently subjective.")
], style={'fontFamily': 'Arial, sans-serif', 'padding': '20px', 'maxWidth': '1200px', 'margin': '0 auto'})
```

## 18. Connect the callback

Connect the control's `value` to the `figure` property of both graphs. The callback function should call the reusable figure function and return its two figures in the same order as the two outputs.



```python

@app.callback(
    Output("stress-graph", "figure"),
    Output("isolation-graph", "figure"),
    Input("industry-dropdown", "value")
)
def update_dashboard(selected_industry):
    return make_dashboard_figures(selected_industry)
```

## 19. Check the callback as a Python function

Call the callback or reusable figure function with at least two valid control values. Confirm that each call returns two Plotly figures without an error.



```python
test_fig1, test_fig2 = update_dashboard("IT")
test_fig1.show()
test_fig2.show()

```
