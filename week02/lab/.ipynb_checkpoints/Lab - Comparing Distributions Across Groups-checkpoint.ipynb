## Step 3: Operation


### Make the borough names human-readable

The source stores borough names in all capital letters. **Title case** capitalizes the first letter of each word and writes the remaining letters in lowercase. For example, `STATEN ISLAND` becomes `Staten Island`.

Use `.str.title()` to convert every value in `requests["borough"]` to title case and save the result back into the same feature.


```python
# Write your code here.
requests["borough"] = requests["borough"].str.title()
```

**What roles do `.str` and `.title()` play in this statement?**

Your answer: `.str` is an accessor that allows you to perform vectorized string operations on an entire pandas Series or Index at once. `.title()`capitalizes the first letter of every word while making the rest lowercase.


### Create a calendar date

Create `requests["date"]` with `requests["created_date"].dt.normalize()`. The `.dt` accessor opens Pandas' datetime tools, and `.normalize()` sets the time in every value to midnight while preserving its calendar date.


```python
# Write your code here.
requests["date"] = requests["created_date"].dt.normalize()
```

### Count requests for each borough and date

Create a DataFrame named `daily_requests` with this sequence:

1. group `requests` by `date` and `borough`;
2. use `.size()` to count rows in every group; and
3. use `.reset_index(name="request_count")` to return a DataFrame whose count column is named `request_count`.

Display the first five rows.


```python
# Write your code here.
daily_requests = requests.groupby(['date', 'borough']).size().reset_index(name="request_count")
daily_requests.head()
```

**What does one row of `daily_requests` represent? How is that different from one row of `requests`?**

Your answer: One row of `daily_requests` represents the total number of requests made in one specific borough on one specific day. Meanwhile, one row of `requests` is a single call or report made. 


**Why did we remove the time before grouping and counting the requests?** As a hint, try repeating the calculation with `created_date` in place of `date` and inspect the result.

Your answer: If we kept the time probably every request would be in its own group since it is likely super rare to have the same call at the exact same time down to the millisecond. 


### Set the borough order

Create `borough_order` containing the five borough names in alphabetical order.


```python
# Write your code here.
borough_order = ['Bronx', 'Brooklyn', 'Manhattan', 'Queens', 'Staten Island']
```

### Calculate the mean daily request count

Create `borough_summary` by grouping `daily_requests` by `borough`, selecting `request_count`, calculating the mean, and using `.reindex(borough_order)`.


```python
# Write your code here.
borough_summary = daily_requests.groupby('borough')['request_count'].mean().reindex(borough_order)
borough_summary
```

**Why is comparing only these means incomplete?**

Your answer: A mean only shows the average, doesn't show the shape of distribution, how the numbers may fluctuate day to day, and whether there are extreme outliers.

## Step 5: Evidence


### Create overlapping histograms

Use `sns.histplot()` with:

- `data=daily_requests`;
- `x="request_count"`;
- `hue="borough"` and `hue_order=borough_order`;
- `element="step"`;
- layered histograms;
- `alpha=0.45`; and
- the `"deep"` palette.

Capitalize the legend title and add a question-focused title and descriptive axis labels. Label the y-axis `Number of days`.


```python
# Write your code here.
sns.histplot(data=daily_requests, 
            x="request_count",
            hue="borough", hue_order=borough_order,
            element="step", 
            multiple="layer",
            alpha=0.45,
            palette="deep")

plt.title("Daily Reported Street Condition Requests by Borough")
plt.xlabel("Requests per day")
plt.ylabel("Number of days")
plt.gca().get_legend().set_title("Borough")
plt.show()
```

**What does the histogram reveal, and what makes the boroughs easy or difficult to compare?**

Your answer: The histogram reveals the spread and overlapping ranges of the daily request counts. You can see the clustering, but this does create a little bit of visual clutter because of the overlapping of the steps. 


### Create a box plot

Put `request_count` on the x-axis and `borough` on the y-axis. Use `borough_order`, the named color `"mediumseagreen"`, and descriptive labels.


```python
# Write your code here.
sns.boxplot(data=daily_requests,
            x='request_count',
            y='borough',
            order=borough_order,
            color="mediumseagreen")
plt.title("Daily Reported Street Condition Requests by Borough")
plt.xlabel('Requests per day')
plt.ylabel('Borough')
plt.show()
```

**What differences among boroughs are easier to see in the box plot than in the histogram?**

Your answer: Just like in the lecture, the box plot makes it incredibly easy to compare the exact medians. It also helps remove visual clutter and it also clearly isolates potential outlier days as individual circles on the right side.


### Create a violin plot

Use the same x feature, y feature, and order. Change only the plot color to `"mediumpurple"` in addition to the required data mappings and order.


```python
# Write your code here.
sns.violinplot(data=daily_requests,
            x='request_count',
            y='borough', 
            order=borough_order,
            color="mediumpurple")
plt.title("Daily Reported Street Condition Requests by Borough")
plt.xlabel('Requests per day')
plt.ylabel('Borough')
plt.show()
```

**What does the violin plot add to the comparison?**

Your answer: It adds a smoothed estimate of the distribution's shape (the density). You can also see where the data bulges, showing where the most common daily counts are clustered. Also, like we discussed in the lecture it really does look like an Amoeba. 


**Which of the three figures most clearly compares daily request counts among boroughs? Explain your choice.**

Your answer: The box plot takes the gold here. The histograms clutter is too messy, while the violin is nice but not as descriptive. The box plot gives a clean, compact view of the center, spread, and outliers, making it instantly obvious how the boroughs stack up against each other. 


## Step 6: Conclusion


**What conclusion about borough differences is supported by the 2025 data?**

Your answer: Brooklyn and Queens experience significantly higher daily 311 service requests compared to the other boroughs. While Staten Island (outside of a few outliers) experiences the least amount of requests.

# Your own analysis

Develop a new research question using the 311 data. Your question must compare a **quantitative feature across categorical groups**. You will probably need to calculate the quantitative feature from the original request records.

Possible grouping features include `descriptor`, month, or day of the week. You may use another meaningful categorical feature that you can create and explain.


## Step 1: Question


**Write a research question that compares a quantitative feature across categorical groups.**

Your answer: How does the typical daily volume of NYC 311 street condition requests differ by the day of the week?


## Step 2: Data


**Which original features will you use, and what will each row of your calculated DataFrame represent?**

Your answer: I will use the created `date` feature to extract the day of the week, and count the total number of requests. Each row in my dataframe will represent a calendar date in 2025, storing the day name and the total number of requests logged citywide on that day. 


## Step 3: Operation

Prepare the grouping feature, calculate the quantitative feature, and save the resulting DataFrame with a meaningful name.


```python
# Write your code here.
#I used geeks for geeks to research this. 
requests['day_of_week'] = requests['date'].dt.day_name(locale='English')

#group by date and day_of_week to get total citywide counts per day
citywide_daily = requests.groupby(['date', 'day_of_week']).size().reset_index(name='total_requests')

#define the days of the week order
order_dow = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
```

**Briefly explain how your code produced the quantitative feature and comparison groups.**

Your answer: I used `.dt.day_name(locale='English')` on the date column to create the categorical grouping feature (the days of the week). Then, I grouped the data by both the exact date and the day of the week, using `.size()` to calculate my quantitative feature: the total number of requests made across the entire city on each specific calendar day. Finally, I just put the days of the week in order. 


## Step 4: Check

Check the calculated data for missing or impossible values, confirm that the intended groups are present, and use an appropriate numerical summary to inspect the distributions.


```python
# Write your code here.
#check that we have 52 or 53 of each day of the week in a year. 
citywide_daily['day_of_week'].value_counts()
```

```python
#inspect the summary for any "impossible" or out of the ordinary issues. 
citywide_daily.groupby('day_of_week')['total_requests'].describe().reindex(order_dow)
```

**What did your checks establish?**

Your answer: The checks showed that each day of the week appeared 52 or 53 times, which matches the number of weeks in a year. There are also no "impossible" or out of the oridnary issues with the data. 


## Step 5: Evidence

Create a histogram, box plot, or violin plot that directly addresses your question. Include a clear title and descriptive axis labels.


```python
# Write your code here.
sns.boxplot(data=citywide_daily,
    x='total_requests',
    y='day_of_week',
    order=order_dow,
    color='crimson')
plt.title("Citywide Daily 311 Street Condition Requests by Day of Week")
plt.xlabel("Total Requests per Day")
plt.ylabel("Day of the Week")
plt.show()
```

**What pattern in the figure provides the strongest evidence for your answer?**

Your answer: The box plot shows a sharp drop off in the number of requests during the weekend. 


## Step 6: Conclusion


**State one conclusion that directly answers your research question.**

Your answer: This shows that the highest number of requests come in during the normal work week, spiking on Tuesday and slowly coming back down as the weekend approaches, then falling off on the weekend.


## Step 7: Limitation


**State one important limitation on how your result should be interpreted.**

Your answer: This analysis only tracks when people reported the problems, not when the street damage actually occurred. The drop in the weekend is likely just the way we live our daily lives, with people seeing and reporting these issues during their daily commute, and then being too busy with their personal lives on the weekend to make a call. This doesn't mean the street issues magically go away on the weekend.
